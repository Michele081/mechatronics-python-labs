```python
# ==========================================
# SIMULATORE DI ROBOT MOBILE (AGV) - LIVELLO 0
# ==========================================

# 1. FUNZIONE: Controlla lo stato dei sensori
def controlla_sensori(distanza_ostacolo, livello_batteria):
    if livello_batteria <= 0:
        return "BATTERIA_SCARICA"
    elif distanza_ostacolo < 20:
        return "OSTACOLO"
    else:
        return "LIBERO"

# 2. VARIABILI: Impostiamo lo stato iniziale del robot
batteria = 4            # Livello di batteria iniziale
posizione_attuale = 0   # Metri percorsi
obiettivo = 3           # Metri da raggiungere per completare la missione

print("--- INIZIO MISSIONE ROBOT ---")

# 3. CICLO WHILE: Il robot si muove finché non arriva a destinazione o finisce la batteria
while posizione_attuale < obiettivo and batteria > 0:
    print(f"\n[Controllo] Posizione: {posizione_attuale}m | Batteria residua: {batteria}%")

    # Per semplicità, simuliamo che davanti al robot ci sia sempre strada libera (50 cm)
    distanza_dal_muro = 50

    # Chiamiamo la funzione che abbiamo creato sopra
    risultato_sensori = controlla_sensori(distanza_dal_muro, batteria)

    # 4. CONDIZIONI (if/elif/else): Decidiamo cosa fare in base ai sensori
    if risultato_sensori == "LIBERO":
        print("-> Azione: Motori attivati. Il robot avanza di 1 metro.")
        posizione_attuale = posizione_attuale + 1  # Avanziamo di 1 metro
        batteria = batteria - 1  #Consumo di batteria dell' 1 %
    elif risultato_sensori == "OSTACOLO":
        print("-> Azione: Ostacolo rilevato! Arresto di sicurezza.")
        break  # Interrompe il ciclo
    elif risultato_sensori == "BATTERIA_SCARICA":
        print("-> Azione: Batteria esaurita! Spegnimento.")
        break

# 5. CONCLUSIONE DELLA MISSIONE
print("\n--- FINE MISSIONE ---")
if posizione_attuale >= obiettivo:
    print("Successo! Il robot ha raggiunto la destinazione finale con successo.")
else:
    print("Missione interrotta prima del traguardo.")
```

```python
import random

def controlla_sensori(distanza_ostacolo, livello_batteria):
    if livello_batteria <= 0:
        return "BATTERIA_SCARICA"
    elif distanza_ostacolo < 20:
        return "OSTACOLO"
    elif livello_batteria == 1:
        # Stampiamo l'avviso ma diciamo al robot che la strada è comunque libera (se non ci sono muri)
        print("ATTENZIONE: Batteria critica, rientro alla base consigliato!")
        return "LIBERO"
    else:
        return "LIBERO"

# Impostiamo lo stato iniziale
batteria = 5
posizione_attuale = 0
obiettivo = 4

print("--- INIZIO MISSIONE DINAMICA ---")

while posizione_attuale < obiettivo and batteria > 0:

    distanza_dal_muro = random.randint(10, 70)

    print(f"\n[Sensore] Distanza misurata dal muro: {distanza_dal_muro} cm")
    print(f"Stato attuale -> Posizione: {posizione_attuale}m | Batteria: {batteria}%")

    risultato_sensori = controlla_sensori(distanza_dal_muro, batteria)

    if risultato_sensori == "LIBERO":
        print("-> Azione: Il robot avanza di 1 metro.")
        posizione_attuale = posizione_attuale + 1
        batteria = batteria - 1

    elif risultato_sensori == "OSTACOLO":
        print("-> Azione: OSTACOLO RILEVATO! Frenata di emergenza.")
        break

    elif risultato_sensori == "BATTERIA_SCARICA":
        print("-> Azione: Batteria esaurita!")
        break

print("\n--- FINE MISSIONE ---")
if posizione_attuale >= obiettivo:
    print("Successo! Il robot ha completato il percorso.")
else:
    print(f"Missione interrotta. Posizione finale: {posizione_attuale}m su {obiettivo}m.")
```

```python
# ==========================================
# MECCATRONICA & LOGISTICA - GESTIONE LISTE DATI
# ==========================================

# 1. Creiamo una lista vuota per memorizzare i dati della missione
cronologia_posizioni = []
temperature_motore = [42.5, 43.0, 45.2, 48.1, 52.0] # Gradi Celsius espressi in numeri decimali

# 2. Simuliamo il robot che avanza e salva le posizioni percorse
for metro in range(1, 6):
    cronologia_posizioni.append(metro) # .append() aggiunge un elemento in coda alla lista
    print(f"Robot avanzato a {metro}m -> Cronologia aggiornata: {cronologia_posizioni}")

print("\n--- ANALISI TELEMETRIA MOTORE ---")

# 3. Analizziamo le temperature registrate dal sensore meccatronico
temperatura_massima = max(temperature_motore) # Trova il valore più alto nella lista
temperatura_media = sum(temperature_motore) / len(temperature_motore) # Calcola la media

print(f"Temperature lette durante la corsa: {temperature_motore}°C")
print(f"Temperatura media motore: {temperatura_media:.1f}°C")
print(f"Picco massimo di temperatura: {temperatura_massima}°C")

# 4. Controllo di sicurezza meccatronico
if temperatura_massima > 50.0:
    print("AVVISO MECCATRONICO: Surriscaldamento rilevato nel motore drive! Ridurre la velocità.")
else:
    print("Stato termico ottimale.")
```

```python
# ==========================================
# MECCATRONICA & LOGISTICA - CONTROLLO ATTIVO
# ==========================================

# 1. Stato iniziale del robot
velocita_robot = 100  # Percentuale di velocità massima
temperature_rilevate = []  # Lista vuota per la telemetria in tempo reale

print(f"Avvio missione. Velocità iniziale: {velocita_robot}%")
print("-" * 40)

# 2. Simuliamo il robot che si muove per 5 "step" temporali
# (Immaginiamo che il motore stia sotto sforzo e la temperatura salga)
for secondo in range(1, 6):

    # Calcoliamo una temperatura simulata che sale ad ogni secondo
    # Al secondo 1 = 41°C, al secondo 2 = 44°C, ..., al secondo 5 = 53°C
    temperatura_attuale = 38 + (secondo * 3)

    # Salviamo la temperatura nella nostra lista (cronologia)
    temperature_rilevate.append(temperatura_attuale)

    print(f"Tempo: {secondo}s | Temperatura motore: {temperatura_attuale}°C | Velocità: {velocita_robot}%")

    # 3. IL SISTEMA DI CONTROLLO MECCATRONICO (Decisione autonoma)
    if temperatura_attuale > 50:
        print("ATTENZIONE: Superata la soglia di sicurezza (50°C)!")
        print("Azione: Attivazione raffreddamento forzato e riduzione velocità del 50%.")

        # Riduciamo la velocità dimezzandola
        velocita_robot = velocita_robot / 2

    print("-" * 40)

# 4. Resoconto finale della telemetria salvata nella lista
print("\n--- RESOCONTO FINALE DELLA MISSIONE ---")
print(f"Cronologia temperature registrate: {temperature_rilevate}°C")
print(f"Temperatura massima registrata: {max(temperature_rilevate)}°C")
print(f"Velocità finale del robot: {velocita_robot}%")
```

```python
# ==========================================
# MECCATRONICA & LOGISTICA - USO DELLE FUNZIONI
# ==========================================

# 1. Definiamo una funzione di controllo termico
def verifica_temperatura(temperatura_attuale, velocita_attuale):
    """
    Questa funzione riceve la temperatura e la velocità,
    controlla i limiti e decide se rallentare il robot.
    """
    limite_sicurezza = 50.0

    if temperatura_attuale > limite_sicurezza:
        print(f"ATTENZIONE: {temperatura_attuale}°C supera il limite di {limite_sicurezza}°C!")
        print("Azione: Riduzione della velocità del 50%.")

        # Dimezziamo la velocità
        nuova_velocita = velocita_attuale / 2
        return nuova_velocita  # Restituisce il nuovo valore calcolato
    else:
        print(f"Temperatura ok ({temperatura_attuale}°C). Velocità invariata.")
        return velocita_attuale  # Restituisce la velocità di prima

# ==========================================
# 2. PROGRAMMA PRINCIPALE (MAIN)
# ==========================================

velocita_robot = 100  # Velocità iniziale
temperature_registrate = [41, 44, 47, 50, 53] # Dati dei sensori

print(f"Avvio missione logistica. Velocità iniziale: {velocita_robot}%\n")

# Simuliamo il passaggio del tempo e usiamo la funzione creata sopra
for secondo, temp in enumerate(temperature_registrate, start=1):
    print(f"--- Secondo {secondo} ---")

    # Chiamiamo la funzione passandole i dati attuali
    # La funzione elabora i dati e ci restituisce la velocità aggiornata
    velocita_robot = verifica_temperatura(temp, velocita_robot)

    print(f"Velocità attuale del robot: {velocita_robot}%\n")

print("Missione completata!")
```
